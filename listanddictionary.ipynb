```python
#listoperations
numbers=[25,12,40,18]

print("Original List")
print(numbers)

numbers.append(50)

print("\nAfter Append")
print(numbers)

numbers.insert(2,100)

print("\nAfter Insert")
print(numbers)

numbers.remove(12)

print("\nAfter Remove")
print(numbers)

numbers.sort()

print("\nSorted List")
print(numbers)

numbers.reverse()

print("\nReverse Order")
print(numbers)
```

```python
#Matrix Addition
A=[[1,2],[3,4]]

B=[[5,6],[7,8]]

C=[]

for i in range(2):

    row=[]

    for j in range(2):

        row.append(A[i][j]+B[i][j])

    C.append(row)

print(C)
```

```python
#Contact Management System
cbook={}
while(True):
    print("\nCONTACT MANAGEMENT SYSTEM")
    print("1.Add Contact")
    print("2.Search Contact")
    print("3.Update Contact")
    print("4.Delete Contact")
    print("5.Display Contacts")
    print("6.Exit")
    choice=int(input("Enter Choice : "))

    if choice==1:
        name=input("Enter Name : ")
        phone=input("Enter Phone : ")
        cbook[name]=phone
        print("Contact Added Successfully")

    elif choice==2:
        name=input("Enter Name : ")
        if name in cbook:
            print("Phone :",cbook[name])
        else:
            print("Contact Not Found")

    elif choice==3:
        name=input("Enter Name to be searched : ")

        if name in cbook:
            phone=input("Enter New Phone : ")
            cbook[name]=phone
            print("Updated Successfully")
        else:
            print("Contact Not Found")

    elif choice==4:
        name=input("Enter Name : ")
        if name in cbook:
            del cbook[name]
            print("Deleted Successfully")
        else:
            print("Contact Not Found")

    elif choice==5:
        print("\nCONTACT LIST")

        for name,phone in cbook.items():
            print(name,"\t",phone)
    elif choice==6:
        print("exiting.........Thank You")
        break
    else:

        print("Invalid Choice")
```
