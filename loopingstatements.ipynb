```python
#for loop
for i in range(5):
    print(i)
```

```python
#while Loop
i=1
while i<=5:
    print(i)
    i=i+1
```

```python
# Factorial using For Loop
num=int(input("Enter Number : "))

fact=1

for i in range(1,num+1):

    fact=fact*i

print("Factorial =",fact)
```

```python
#Factorial using While Loop
num=int(input("Enter Number : "))

fact=1

i=1

while i<=num:

    fact=fact*i

    i=i+1

print("Factorial =",fact)
```

```python
#Demonstration of break
for i in range(1,11):
    if i==6:
        break
    print(i)
```

```python
#Demonstration of continue
for i in range(1,11):
    if i==6:
        continue
    print(i)
```

```python
#Demonstration of pass
for i in range(5):
    pass
print("Loop Finished")
```

```python
#Demonstration of else with loop
for i in range(1,6):
    print(i)
else:
    print("Loop Executed Successfully")
```

```python
#"Prime Number
num=int(input("Enter Number : "))
count=0
for i in range(1,num+1):
    if num%i==0:
        count+=1
if count==2:
    print("Prime")
else:
    print("Not Prime")
```

```python
#Number Palindrome
num=int(input("Enter Number : "))
temp=num
rev=0
while temp>0:
    digit=temp%10
    rev=rev*10+digit
    temp=temp//10
if rev==num:
    print("Palindrome")
else:
    print("Not Palindrome")
```
